```
Code a savings goal tracker. Ask the user:
- "what are you saving up for?"
- "how much do you need to save up?"
- "how much have you already saved?"
- "when do you need to complete the goal by?"
Create an action plan and feedback
alculates the current progress = 2-1
- calculates how much time they have left in months = 3 - current date
- calcultes the interval of how much they should save up each moth leading up to 3 (the month they need to complete the goal by)
```

````python
# trial 1

from datetime import datetime
from dateutil.relativedelta import relativedelta

print("Welcome to your Savings Goal Tracker!")

# 1. Ask for user inputs
goal_name = input("What are you saving up for? ")

while True:
    try:
        target_amount_str = input("How much do you need to save up? ")
        target_amount = float(target_amount_str)
        if target_amount < 0:
            print("Target amount cannot be negative. Please enter a positive number.")
        else:
            break
    except ValueError:
        print("Invalid input. Please enter a numerical value for the target amount.")

while True:
    try:
        saved_amount_str = input("How much have you already saved? ")
        saved_amount = float(saved_amount_str)
        if saved_amount < 0:
            print("Saved amount cannot be negative. Please enter a positive number.")
        else:
            break
    except ValueError:
        print("Invalid input. Please enter a numerical value for the amount already saved.")

while True:
    target_date_str = input("When do you need to complete the goal by? (YYYY-MM-DD) ")
    try:
        target_date = datetime.strptime(target_date_str, '%Y-%m-%d')
        break
    except ValueError:
        print("Invalid date format. Please use YYYY-MM-DD.")

# 2. Perform calculations
current_date = datetime.now()

remaining_amount = target_amount - saved_amount

time_left_delta = relativedelta(target_date, current_date)

# Calculate months left, rounding to the nearest whole number
# If target_date is in the past, time_left_months should be 0
if target_date < current_date:
    time_left_months = 0
elif time_left_delta.years == 0 and time_left_delta.months == 0 and time_left_delta.days < 31:
    # If less than a month left (and not exactly a month or more), consider it 0 months for monthly calculation
    time_left_months = 0
else:
    # Approximate total months, considering full months and a portion of the last month
    total_months = time_left_delta.years * 12 + time_left_delta.months
    if time_left_delta.days > 0:
        total_months += 1 # Add an extra month if there are remaining days
    time_left_months = max(0, total_months)


monthly_saving_needed = 0.0 # Default to 0
if remaining_amount <= 0: # Already saved enough or more
    monthly_saving_needed = 0
elif time_left_months > 0:
    monthly_saving_needed = remaining_amount / time_left_months

# 3. Print action plan and feedback
print("\n--- Savings Goal Action Plan ---")
print(f"Goal: {goal_name}")
print(f"Target Amount: ${target_amount:,.2f}")
print(f"Amount Already Saved: ${saved_amount:,.2f}")
print(f"Remaining Amount to Save: ${max(0, remaining_amount):,.2f}")
print(f"Target Completion Date: {target_date.strftime('%Y-%m-%d')}")

if remaining_amount <= 0:
    print("Congratulations! You have already reached or exceeded your savings goal!")
elif time_left_months == 0:
    print(f"You have very little time left (less than a month or the deadline has passed) to save ${remaining_amount:,.2f}.")
    print("It's not possible to calculate a monthly saving amount based on whole months.")
else:
    print(f"Time Left to Save (in months): {time_left_months}")
    print(f"Recommended Monthly Saving: ${monthly_saving_needed:,.2f}")

print("\nGood luck with your savings goal!")
````
