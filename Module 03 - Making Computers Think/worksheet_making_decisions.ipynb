### Activity 3: Debug the broken classifier

Here is some code that looks fine but misbehaves. Do not fix it by guessing. Trace the logic first.

Run the cell below. It should give "Good Standing" for a GPA of 3.2. Does it?

````python
student_gpa = 3.2

if student_gpa >= 2.0:
    status = "Probation"
elif student_gpa >= 3.0:
    status = "Good Standing"
elif student_gpa >= 3.5:
    status = "Dean's List"
else:
    status = "Failing"

print(f"Status: {status}")
````

**Trace it, then fix it.**

- When `student_gpa = 3.2`, is `3.2 >= 2.0` true? What happens the moment a condition is true?
- Why does the order of the conditions matter here?

> **Your analysis** (double-click to edit):

```
The problem is: For a GPA of 3.2, the code incorrectly outputs 'Probation' instead of 'Good Standing'.

It happens because: The 'if student_gpa >= 2.0' condition is evaluated first. Since 3.2 is greater than or equal to 2.0, this condition is met, and the `status` is set to 'Probation'. Python's `if/elif/else` structure executes the first true condition and then skips the rest of the `elif` blocks. This means the more specific conditions for 'Good Standing' (GPA >= 3.0) and 'Dean's List' (GPA >= 3.5) are never reached for a GPA of 3.2.

The fix: The conditions must be ordered from most specific (highest GPA range) to least specific (lowest GPA range) to ensure that the correct status is assigned.
```

Now write your corrected version below.

````python
# Your fixed version here
student_gpa = 3.2

if student_gpa >= 3.5:
    status = "Dean's List"
elif student_gpa >= 3.0:
    status = "Good Standing"
elif student_gpa >= 2.0:
    status = "Probation"
else:
    status = "Failing"

print(f"Status: {status}")
````

### Implementing the 'Night Owl' Achievement Logic

````python
def check_night_owl_achievement(played_hours_list):
    """
    Checks if the 'Night Owl' achievement is earned.

    Requirements:
    - Playing late (between 10 PM and 4 AM, inclusive) for 3 consecutive days.

    Args:
        played_hours_list (list): A list of integers representing the hour
                                  (0-23) the player played on consecutive days.
                                  For example, [23, 1, 2] means playing at 11 PM
                                  on day 1, 1 AM on day 2, and 2 AM on day 3.

    Returns:
        bool: True if the achievement is earned, False otherwise.
    """
    required_consecutive_days = 3

    if len(played_hours_list) < required_consecutive_days:
        return False # Not enough days to meet the '3 days in a row' requirement

    consecutive_late_days = 0
    for hour in played_hours_list:
        # Check if the hour is between 10 PM (22) and 4 AM (4)
        # Note: 4 AM is included, so hour <= 4 is correct.
        # For 10 PM, 22 <= hour is correct. For 11 PM, 23 <= hour is correct.
        if (hour >= 22 and hour <= 23) or (hour >= 0 and hour <= 4):
            consecutive_late_days += 1
            if consecutive_late_days >= required_consecutive_days:
                return True
        else:
            # If a day is not played late, reset the consecutive count
            consecutive_late_days = 0

    return False # If loop finishes and condition not met


# --- Test Cases ---
print("Night Owl Achievement Test Cases:")

# Player 1: Clearly earns it (3 consecutive late nights)
player1_hours = [23, 1, 3]
print(f"Player 1 hours: {player1_hours} -> Achievement: {check_night_owl_achievement(player1_hours)}")

# Player 2: Does not earn it (not enough consecutive days)
player2_hours = [23, 1]
print(f"Player 2 hours: {player2_hours} -> Achievement: {check_night_owl_achievement(player2_hours)}")

# Player 3: Does not earn it (breaks the streak)
player3_hours = [23, 1, 10, 2]
print(f"Player 3 hours: {player3_hours} -> Achievement: {check_night_owl_achievement(player3_hours)}")

# Player 4: Earns it after a break
player4_hours = [15, 23, 0, 4]
print(f"Player 4 hours: {player4_hours} -> Achievement: {check_night_owl_achievement(player4_hours)}")

# Player 5: Borderline case (exactly 3 late nights)
player5_hours = [22, 23, 0]
print(f"Player 5 hours: {player5_hours} -> Achievement: {check_night_owl_achievement(player5_hours)}")

# Player 6: Not late enough (boundary condition)
player6_hours = [5, 23, 1]
print(f"Player 6 hours: {player6_hours} -> Achievement: {check_night_owl_achievement(player6_hours)}")
````

> **Your criteria** (double-click to edit):

```
Factor 1: Test Scores
Factor 2: Extra Activities
Factor 3: Special Circumstances
Decision logic:
- Test Scores below 60%, No Extra Activities, No Special Circumstances, if conditions meet = denied
- Test Scores below 60%, Extra Activities = volunteering OR portfolio, No Special Circumstances, if condistions meet = considered
- Test Scores below 60%, No Extra Activities,  Special Circumstances = disability OR financial hardship, if conditions meet = considered
- Test Scores above 60%, No Extra Activities, No Special Circumstances, if conditions meet = passed
- Test Scores above 80%, No Extra Activities, Special Circumstances = disability OR financial hardship, if conditions meet = scholarship offered
```

````python
# Your admissions decision system
test_scores = float(input("Test Scores: "))
extra_activities = input("Extra Activities: ")
special_circumstances = input("Special Circumstances: ")

if test_scores < 60 and extra_activities == "No" and special_circumstances == "No":
    print("Denied")

elif test_scores < 60 and (extra_activities == "Yes" or special_circumstances == "Yes"):
    print("considered")

elif test_scores >= 60 and extra_activities == "No" and special_circumstances == "No":
    print("Passed")

elif test_scores >= 80 and (extra_activities == "Yes" or special_circumstances == "Yes"):
    print("Scholarship Offered")

# Tip: for "at least 2 of 3 conditions true", try counting how many are True.
````

````python
# Your admissions decision system with AI
test_scores = float(input("Test Scores: "))
extra_activities_input = input("Enter Extra Activities (e.g., 'volunteering', 'portfolio', or 'none'): ").lower()
special_circumstances_input = input("Enter Special Circumstances (e.g., 'disability', 'financial hardship', or 'none'): ").lower()

# Helper booleans to check for specific activities/circumstances
has_volunteering_or_portfolio = "volunteering" in extra_activities_input or "portfolio" in extra_activities_input
has_disability_or_financial_hardship = "disability" in special_circumstances_input or "financial hardship" in special_circumstances_input

decision = "Undetermined" # Initialize decision

# Decision logic based on the specified rules, ordered from most specific/highest tier

# Rule: Test Scores above 80%, No Extra Activities, Special Circumstances = disability OR financial hardship = scholarship offered
# Note: 'No Extra Activities' means specifically not having volunteering or portfolio.
if test_scores >= 80 and not has_volunteering_or_portfolio and has_disability_or_financial_hardship:
    decision = "Scholarship Offered"
# Rule: Test Scores above 60% (and not a scholarship case), No Extra Activities, No Special Circumstances = passed
elif test_scores >= 60:
    if not has_volunteering_or_portfolio and not has_disability_or_financial_hardship:
        decision = "Passed"
    else:
        # Implicitly, if score >= 60 and not a scholarship, and has other factors, still 'Passed'
        # This covers cases like 'Test Scores 70%, volunteering, no special circumstances'
        decision = "Passed"
# Rule: Test Scores below 60%, Extra Activities OR Special Circumstances (specific types) = considered
elif test_scores < 60:
    if has_volunteering_or_portfolio or has_disability_or_financial_hardship:
        decision = "Considered"
    # Rule: Test Scores below 60%, No Extra Activities, No Special Circumstances = denied
    else:
        decision = "Denied"

print(f"Decision: {decision}")
````

## Key takeaways

**Computational thinking**
- Understand the problem before you code.
- Break decisions into clear rules.
- Test your logic with examples first.
- In an if/elif chain, order matters. Put the most specific conditions first.
- Use brackets to make complex conditions say what you mean. `and` binds tighter than `or`.

- AI is at its best when your requirements are clear.
- Use it to build your ideas, not to think for you.
