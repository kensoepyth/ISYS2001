# Dictionaries: keeping related information together

Over the last few weeks we have been breaking problems into small jobs and giving
each one a name with a function. Today we look at a way to keep related pieces of
information together, so that we can look something up by name rather than by its
position in a list.

This is a guided worksheet. We will run each cell together and talk through it as
we go. Keep your attention on *when* you would reach for a dictionary and *why*, not
just on the syntax. The syntax is short. The judgement is the part worth having.

## Our example: a month of spending

Here is a small record of what someone spent last month, grouped by category. The
category names are the keys, and the amounts are the values. This is a dictionary.

```python
spending = {
    'rent': 1200,
    'groceries': 480,
    'transport': 150,
    'entertainment': 90,
}

spending
```

## Looking something up

Because the keys are meaningful names, we can ask for exactly the piece we want.

```python
print(spending['groceries'])
```

## What happens if the key is not there?

Have a guess at what this next cell will do before we run it.

```python
# We never recorded anything under 'savings'.
print(spending['savings'])
```

## A safer way to look up: `.get()`

Asking for a key that does not exist stops the whole program. In a real program that
is rarely what we want. The `.get()` method looks something up without crashing: it
hands back nothing, or a sensible default we choose, when the key is missing.

```python
print(spending.get('savings'))          # nothing, but no crash
print(spending.get('savings', 0))       # a sensible default instead
print(spending.get('groceries', 0))     # still works for keys that do exist
```

## Adding and changing entries

A dictionary is not fixed once we make it. We can add a new category, or correct one
that is already there, using the same square-bracket notation.

```python
spending['savings'] = 200      # add a category we did not have before
spending['transport'] = 175    # correct an amount that was already there

spending
```

## Going through everything

When we want to walk through the whole dictionary, `.items()` gives us the key and
the value together on each pass.

```python
for category, amount in spending.items():
    print(f"{category}: ${amount}")
```

## Summarising across the whole record

Once the data is in one place, questions about all of it become easy to ask. Here is
the total, and the single category that took the most.

```python
total = sum(spending.values())
print(f"Total spent: ${total}")

biggest = max(spending, key=spending.get)
print(f"Biggest category: {biggest}")
```

## Wrapping a job up in a function

This is the same move we made with functions a couple of weeks ago. Notice that this
function *returns* the total rather than printing it, so we are free to use the answer
somewhere else.

```python
def total_spending(record):
    return sum(record.values())

print(total_spending(spending))
```
